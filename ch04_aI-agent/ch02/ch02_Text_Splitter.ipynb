```python
import urllib.request
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
urllib.request.urlretrieve(
    "https://raw.githubusercontent.com/lovit/soynlp/master/tutorials/2016-10-20.txt",
    filename="2016-10-20.txt")
```

```python
with open("2016-10-20.txt", encoding="utf-8") as f: file=f.read()
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500, chunk_overlap=0)
```

```python
texts = text_splitter.create_documents([file])
print('분할된 청크의 수:', len(texts))
```

```python
texts[1]
```

```python
print("1번 청크의 길이:", len(texts[1].page_content))
print("2번 청크의 길이:", len(texts[2].page_content))
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500, chunk_overlap=50)
texts = text_splitter.create_documents([file])
print('분할된 청크의 수:', len(texts))
```
